# 3.9.1. El Paquete Requests
En Python se puede instalar un paquete llamado `Requests` para realizar solicitudes HTTP, es muy amigable para los desarrolladores. 

Aunque se pueden utilizar todos los métodos de solicitud HTTP (por ejemplo, post, put, head, etc.), en la mayoría de los casos utilizarás el método get mediante requests.get(), al cual le pasarás la URL como una cadena de texto.

````python
import requests

url = 'https://static.webucator.com/media/public/documents/hrleaders.html'
````

````python
r = requests.get(url)
content = r.text
print(content[:225]) # imprime los primeros 225 caracteres del contenido HTML
````

# El parámetero headers 
permite enviar información adicional al servidor, 
como el tipo de navegador que se está utilizando. Esto puede ser útil para evitar bloqueos o restricciones 
en algunos sitios web.

user-agent es un encabezado HTTP que identifica el navegador y el sistema operativo del cliente que realiza la solicitud.
my-app/0.0.1 es un valor personalizado que indica que la solicitud proviene de una aplicación específica, 
en lugar de un navegador web común.


````python
r = requests.get(url, headers={'user-agent': 'my-app/0.0.1'})
content = r.text
print(content[:225]) 
````

### Consultar una página y revisar si respondió correctamente

````python
print(r.status_code)
````

significa que la solicitud fue exitosa.

Esto te permite introducir códigos HTTP básicos:

* `200`: correcto.
* `404`: recurso no encontrado.
* `403`: acceso prohibido.
* `500`: error del servidor.


### Obtener información de la respuesta


````python
print(r.headers["Content-Type"])
````

una respuesta HTTP no contiene solamente el contenido; también incluye **metadatos en los encabezados

### Consultar una API que devuelve JSON
`requests` **no solamente sirve para descargar páginas HTML**.


````python
urlConJson = "https://jsonplaceholder.typicode.com/users/1"

respuesta = requests.get(urlConJson)

datos = respuesta.json()

print(datos)
````

Ya se puede acceder directamente:

````python
print(datos["name"])
print(datos["email"])
````

### Enviar parámetros en una URL

En vez de construir manualmente algo como:

```text
https://jsonplaceholder.typicode.com/comments?postId=1
```

puedes escribir:


````python
parametros = {
    "postId": 1
}

respuesta = requests.get(urlConJson, params=parametros)

print(respuesta.url)
datos = respuesta.json()

print(datos["name"])
````

### Trabajar con encabezados HTTP

````python
import requests

url = "https://httpbin.org/headers"

headers = {
    "User-Agent": "CursoPython/1.0"
}

respuesta = requests.get(url, headers=headers)

print(respuesta.json())
````

Podemos enviar información adicional junto con la solicitud:

```text
GET /recurso
User-Agent: CursoPython/1.0
```

````python
requests.get(
    url,
    headers={"user-agent": "my-app/0.0.1"}
)
````

### Manejar errores de manera básica

En lugar de asumir que todo funcionó:


````python
import requests

url = "https://www.python.org"

respuesta = requests.get(url)

if respuesta.status_code == 200:
    print("Solicitud correcta")
else:
    print("Ocurrió un problema")
````

versión orientada al uso real

````python
url = "https://www.python.org"

respuesta = requests.get(url)

respuesta.raise_for_status()

print("Solicitud correcta")
````

`raise_for_status()` genera una excepción si ocurre un error HTTP.

### Utilizar `timeout`

> Espera hasta 5 segundos antes de considerar que existe un problema con la solicitud.

En código real es buena práctica no dejar una solicitud esperando indefinidamente.


````python
import requests

url = "https://www.python.org"

respuesta = requests.get(url, timeout=5)

print(respuesta.status_code)
````

### Descargar un archivo

````python
url = "https://www.python.org/static/img/python-logo.png"

respuesta = requests.get(url)

with open("python-logo.png", "wb") as archivo:
    archivo.write(respuesta.content) # contenido binario.
````


| Propiedad          | Uso típico                            |
| ------------------ | ------------------------------------- |
| `response.text`    | HTML, XML, texto                      |
| `response.json()`  | JSON                                  |
| `response.content` | Imágenes, PDF, ZIP, archivos binarios |

# 3.9.2. HTML
Los datos en las páginas web se almacenan en HTML, un lenguaje de marcado relativamente simple. Los elementos de una página HTML se marcan mediante etiquetas (tags). Por ejemplo, un párrafo comienza con una etiqueta de apertura `<p>` y termina con una etiqueta de cierre `</p>`. La página mostrada presenta registros de jonrones en una tabla. Cada registro se encuentra en una fila de tabla marcada de la siguiente manera:

```html
<tr class="steroids-era">
    <td>1</td>
    <td>Mark McGwire</td>
    <td>70</td>
    <td>1998</td>
    <td>1963-10-01</td>
</tr>
<a url="1963-10-01" />
```
- Las filas de una tabla comienzan con una etiqueta de apertura `<tr>` y terminan con una etiqueta de cierre `</tr>`.
- Dentro de las filas de una tabla, las celdas de datos comienzan con una etiqueta de apertura `<td>` y terminan con una etiqueta de cierre `</td>` 
- Las celdas de encabezado comienzan con una etiqueta de apertura `<th>` y terminan con una etiqueta de cierre `</th>`.
- Los atributos normalmente vienen en pares nombre-valor.
- los atributos solo aparecen en la etiqueta de apertura, de esta forma:

```html
<tagname att1="value" att2="value">Contenido del elemento</tagname>
```

Algunas etiquetas tienen un atributo class":
```html
<tr class="steroids-era">
```

Es común buscar elementos por nombre de etiqueta y por clase.

Escribir código para extraer datos utilizando manipulación de cadenas y expresiones regulares sería bastante complicado.
